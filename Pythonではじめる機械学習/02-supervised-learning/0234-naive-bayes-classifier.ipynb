```python
# ナイーブベイズクラス分類器
# 線形モデルよりも訓練が高速だが、汎化性能はわずかに劣る場合が多い
# クラスに対する統計値を個々の特徴量ごとに集めてパラメータを学習する
# sklearnには、GaussianNB, BernoulliNB, MultinomialNB の3種の分類器が実装されている
import numpy as np

X = np.array([
    [0, 1, 0, 1],
    [1, 0, 1, 1],
    [0, 0, 0, 1],
    [1, 0, 1, 0],
])
y = np.array([0, 1, 0, 1])

counts = {}
for label in np.unique(y):
    counts[label] = X[y == label].sum(axis=0)
print("Feature counts:\n{}".format(counts))
```
