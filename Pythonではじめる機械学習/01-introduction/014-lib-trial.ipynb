```python
import numpy as np

x = np.array([[1,2,3], [4,5,6]])
print("x:\n{}".format(x))
```

```python
from scipy import sparse

# 階数4の単位行列を作成する（Numpy配列）
# 単位行列は I で表現されることが多い
eye = np.eye(4)
print("Numpy Array:\n{}".format(eye))
```

```python
# Numpy配列を、ScipyのCSR形式の疎行列に変換する（軽量なCSR形式への変換）
sparse_matrix = sparse.csr_matrix(eye)
print("Scipy sparse CSR matrix:\n{}".format(sparse_matrix))
```

```python
data = np.ones(4)          # [1. 1. 1. 1.]
row_indices = np.arange(4) # [0 1 2 3]
col_indices = np.arange(4) # [0 1 2 3]

# 疎行列を作成する材料が揃ったので、COOからCSRに変換
eye_coo = sparse.coo_matrix((data, (row_indices, col_indices)))
print("COO representation: \n{}".format(eye_coo))
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

# -10 から 10 までを 100 ステップに区切った配列として作成
x = np.linspace(-10, 10, 100)
y = np.sin(x)
plt.plot(x, y, marker=".")
```

```python
import pandas as pd

data = {
    'Name': ["John", "Anna", "Peter", "Linda"],
    'Location': ["New York", "Paris", "Berlin", "London"],
    'Age': [24, 13, 53, 33]
}
# 表形式のデータを扱うための入れ物にデータを入れる
data_pandas = pd.DataFrame(data)
# Jupyter が用意している関数で、裏では IPython が動いている
display(data_pandas)
# SQL の Where 句のような絞り込みも可能
display(data_pandas[data_pandas.Age > 30])
```
