```python
# パーセプトロン
# 複数の信号を入力として受け取り、1つの信号を出力する
# 重みづけされた総和が閾値を超えた場合に1、そうでなければ0を出力
# パラメータと閾値を調整すれば、ANDゲートやNANDゲート、ORゲートなどの論理回路を構築することができる
```

```python
# 重みと閾値による AND ゲート
def AND_TMP(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.7
    sum = x1 * w1 + x2 * w2 
    if sum <= theta:
        return 0
    elif sum > theta:
        return 1

AND_TMP(0, 0), AND_TMP(1, 0), AND_TMP(0, 1), AND_TMP(1, 1)
```

```python
# 重みとバイアスによる AND ゲート
# 重みとバイアスの設定値を変更することで、 NAND ゲートや OR ゲートを作成することができる
import numpy as np

def AND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = - 0.7
    sum = np.sum(w * x) + b
    if sum <= 0:
        return 0
    elif sum > 0:
        return 1

AND(0, 0), AND(1, 0), AND(0, 1), AND(1, 1)
```

```python
def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7
    sum = np.sum(w * x) + b
    if sum <= 0:
        return 0
    elif sum > 0:
        return 1

NAND(0, 0), NAND(1, 0), NAND(0, 1), NAND(1, 1)
```

```python
def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = - 0.3
    sum = np.sum(w * x) + b
    if sum <= 0:
        return 0
    elif sum > 0:
        return 1

OR(0, 0), OR(1, 0), OR(0, 1), OR(1, 1)
```

```python
# 多層パーセプトロン
# XOR ゲート（排他的論理和）については、AND, NAND, OR ゲートを組み合わせて表現する
def XOR(x1, x2):
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    y = AND(s1, s2)
    return y

XOR(0, 0), XOR(1, 0), XOR(0, 1), XOR(1, 1)
```
