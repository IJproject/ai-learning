```python
# CNN（Convolutional Neural Network, 畳み込みニューラルネットワーク）
# Affine-ReLU → Conv-ReLU-Pooling のように構成したネットワーク
# ただし出力に近い層では、 Affine-ReLU のままの組み合わせが使用される

# Affineのような、前の層のすべてのニューロンが
# 次の層のすべてのニューロンとつながっている層のことを、全結合層という
# 全結合層の欠点として、3次元データを1次元に均して入力する必要があるために、
# 空間的な形状に関する情報を消すことになってしまう
# →CNNでは、形状を維持した状態で、次の層に出力していく
```

```python
# 畳み込み層
# CNNでは、畳み込み層への入出力データのことを、特徴マップ（feature map）という

# 畳み込み演算
# 入力データに対して、フィルターのウィンドウを一定の間隔でスライドさせながら適用していく

# パディング
# 畳み込み演算を行う前に、入力データの周囲に固定データを埋め込むことがある
# 固定データは、0が入った空のピクセルなど
# 出力サイズを調整することが大きな目的

# ストライド
# フィルターを適用する位置の間隔のこと

# 3次元データの畳み込み演算
# 奥行き方向（チャンネル方向）に特徴マップが増えているような状態
# チャンネルごとにフィルターの畳み込みを行い、それらの結果を加算して一つの出力を得る
# フィルターを N 個用意すれば、出力特徴マップのチャンネル数を N 個にすることが可能
```

```python
# プーリング層
# 縦・横方向の空間を小さくする演算（畳み込みと似たような操作を行う）
# Maxプーリング、Averageプーリングなどがある
# 一般にウィンドウサイズと、ストライドは同じ値に設定する
```
