```python
# ニューラルネットワーク
# 先頭の列を入力層、中間の層を中間層（隠れ層）、最後の列を出力層と呼ぶ
# 入力はニューロンの値だけでなく、バイアスも入力値として扱う
```

$$
y = h(b + w_1 x_1 + w_2 x_2)
$$

$$
h(x) =
\begin{cases}
0 & (x \le 0) \\
1 & (x > 0)
\end{cases}
$$

```python
# 上記の h(x) を活性化関数という
# 層と層の間で「曲げる」（非線形にする）役割を担っている
# 活性化関数には、非線形関数を用いる必要がある
```

```python
# ステップ関数（階段関数）
# 入力が配列になるケースも想定した実装
import matplotlib.pylab as plt
import numpy as np 

def step_function(x):
    y = x > 0
    return y.astype(int) # 各要素の値を Boolean から Integer に変換する

x = np.arange(-5.0, 5.0, 0.1)
y = step_function(x)
plt.plot(x, y)
plt.ylim(-0.1, 1.1)
plt.show()
```

```python
# シグモイド関数
# ニューラルネットワークでよく用いられる活性化関数の1つ
```

$$
h(x) = \frac{1}{1 + \exp(-x)}
$$

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x)) 

x = np.arange(-5.0, 5.0, 0.1)
y = sigmoid(x)
plt.plot(x, y)
plt.ylim(-0.1, 1.1)
plt.show()
```

```python
# ３層ニューラルネットワークの実装
```

$$
A^{(1)} = XW^{(1)} + B^{(1)}
$$

```python
X = np.array([1.0, 0.5])
W1 = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
B1 = np.array([0.1, 0.2, 0.3])

print(X.shape)
print(W1.shape)
print(B1.shape)

A1 = np.dot(X, W1) + B1
print(A1, A1.shape)

Z1 = sigmoid(A1)
print(Z1, Z1.shape)
```

```python
# 出力層での活性化関数は、一般的に隠れ層とは異なる関数を使用することが多い
# 出力層の活性化関数については、回帰問題では恒等関数、2クラス分類ではシグモイド関数、多クラス分類ではソフトマックス関数を使うことが多い
```

```python
def init_network():
    network = {}
    network['W1'] = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
    network['b1'] = np.array([0.1, 0.2, 0.3])
    network['W2'] = np.array([[0.1, 0.4], [0.2, 0.5], [0.3, 0.6]])
    network['b2'] = np.array([0.1, 0.2])
    network['W3'] = np.array([[0.1, 0.3], [0.2, 0.4]])
    network['b3'] = np.array([0.1, 0.2])
    return network

def identity_function(x):
    return x

# 順伝播
def forward(network, x):
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']

    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)
    a3 = np.dot(z2, W3) + b3
    z3 = identity_function(a3)
    return z3

network = init_network()
x = np.array([1.0, 0.5])
y = forward(network, x)
print(y)
```

```python
# ソフトマックス関数
```

$$
y_k = \frac{\exp(a_k)}{\sum_{i=1}^{n} \exp(a_i)}
$$

$$
\begin{aligned}
y_k &= \frac{\exp(a_k)}{\sum_{i=1}^{n} \exp(a_i)} \\
&= \frac{C \exp(a_k)}{C \sum_{i=1}^{n} \exp(a_i)} \\
&= \frac{\exp(a_k + \log C)}{\sum_{i=1}^{n} \exp(a_i + \log C)} \\
&= \frac{\exp(a_k + C')}{\sum_{i=1}^{n} \exp(a_i + C')}
\end{aligned}
$$

```python
def softmax(a):
    c = np.max(a)
    exp_a = np.exp(a - c) # オーバーフロー対策
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a
    return y

a = np.array([0.3, 2.9, 4.0])
y = softmax(a)
print(y)
print(np.sum(y)) # 総和が1であることは、ソフトマックス関数の特徴（→統計的な知見を得ることが可能）、値の大小関係が変わることもない
```
